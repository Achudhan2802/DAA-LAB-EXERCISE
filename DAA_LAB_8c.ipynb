```python
import random
import math
from itertools import permutations

# -----------------------------
# Generate 10 Random Cities
# -----------------------------
random.seed(42)
n = 10

cities = [(random.randint(0, 100), random.randint(0, 100))
          for _ in range(n)]

# Distance Matrix
dist = [[0] * n for _ in range(n)]

for i in range(n):
    for j in range(n):
        if i != j:
            x1, y1 = cities[i]
            x2, y2 = cities[j]
            dist[i][j] = round(math.hypot(x1 - x2, y1 - y2), 2)


# -----------------------------
# Tour Cost
# -----------------------------
def tour_cost(tour):
    cost = 0
    for i in range(len(tour) - 1):
        cost += dist[tour[i]][tour[i + 1]]
    return round(cost, 2)


# -----------------------------
# Nearest Neighbour
# -----------------------------
def nearest_neighbour():
    visited = [False] * n
    tour = [0]
    visited[0] = True
    current = 0

    while len(tour) < n:
        nearest = None
        best = float('inf')

        for city in range(n):
            if not visited[city] and dist[current][city] < best:
                best = dist[current][city]
                nearest = city

        tour.append(nearest)
        visited[nearest] = True
        current = nearest

    tour.append(0)
    return tour


# -----------------------------
# 2-Opt Improvement
# -----------------------------
def two_opt(tour):
    improved = True

    while improved:
        improved = False

        for i in range(1, len(tour) - 2):
            for j in range(i + 1, len(tour) - 1):

                if j - i == 1:
                    continue

                new_tour = tour[:]
                new_tour[i:j] = reversed(tour[i:j])

                if tour_cost(new_tour) < tour_cost(tour):
                    tour = new_tour
                    improved = True

    return tour


# -----------------------------
# Brute Force Optimal
# -----------------------------
def brute_force():
    best_cost = float('inf')
    best_tour = None

    for perm in permutations(range(1, n)):
        tour = [0] + list(perm) + [0]
        cost = tour_cost(tour)

        if cost < best_cost:
            best_cost = cost
            best_tour = tour

    return best_tour, best_cost


# -----------------------------
# Run Algorithms
# -----------------------------
nn_tour = nearest_neighbour()
nn_cost = tour_cost(nn_tour)

opt_tour = two_opt(nn_tour.copy())
opt2_cost = tour_cost(opt_tour)

best_tour, best_cost = brute_force()

# -----------------------------
# Results
# -----------------------------
print("City Coordinates:")
for i, c in enumerate(cities):
    print(f"City {i}: {c}")

print("\nNearest Neighbour Tour:")
print(nn_tour)
print("Cost =", nn_cost)

print("\n2-Opt Improved Tour:")
print(opt_tour)
print("Cost =", opt2_cost)

print("\nOptimal Tour (Brute Force):")
print(best_tour)
print("Cost =", best_cost)

print("\nComparison")
print("Nearest Neighbour Difference =", round(nn_cost - best_cost, 2))
print("2-Opt Difference =", round(opt2_cost - best_cost, 2))
```
