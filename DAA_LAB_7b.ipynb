```python
def solve_n_queens_bitmask(n):
    solutions = 0
    backtracks = 0

    def backtrack(row, cols, diag1, diag2):
        nonlocal solutions, backtracks

        if row == n:
            solutions += 1
            return

        # Bits representing available positions
        available = ((1 << n) - 1) & ~(cols | diag1 | diag2)

        while available:
            # Pick the rightmost available position
            pos = available & -available
            available -= pos

            backtrack(
                row + 1,
                cols | pos,
                (diag1 | pos) << 1,
                (diag2 | pos) >> 1
            )

            backtracks += 1

    backtrack(0, 0, 0, 0)
    return solutions, backtracks


# Main Program
n = 12
solutions, backtracks = solve_n_queens_bitmask(n)

print("N =", n)
print("Number of Solutions =", solutions)
print("Total Backtracks =", backtracks)
```
