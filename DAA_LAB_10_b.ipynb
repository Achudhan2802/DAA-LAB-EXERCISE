```python
import random
import sys

sys.setrecursionlimit(20000)


# ---------------- Comparison Counter ----------------
class Counter:
    def __init__(self):
        self.comparisons = 0


# ---------------- Deterministic Quick Sort ----------------
def deterministic_quick_sort(arr, low, high, counter):
    if low < high:
        p = partition_first(arr, low, high, counter)
        deterministic_quick_sort(arr, low, p - 1, counter)
        deterministic_quick_sort(arr, p + 1, high, counter)


def partition_first(arr, low, high, counter):
    pivot = arr[low]
    i = low + 1

    for j in range(low + 1, high + 1):
        counter.comparisons += 1
        if arr[j] < pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[low], arr[i - 1] = arr[i - 1], arr[low]
    return i - 1


# ---------------- Randomized Quick Sort ----------------
def randomized_quick_sort(arr, low, high, counter):
    if low < high:
        p = random_partition(arr, low, high, counter)
        randomized_quick_sort(arr, low, p - 1, counter)
        randomized_quick_sort(arr, p + 1, high, counter)


def random_partition(arr, low, high, counter):
    rand_index = random.randint(low, high)
    arr[low], arr[rand_index] = arr[rand_index], arr[low]

    return partition_first(arr, low, high, counter)


# ---------------- Median-of-Three Quick Sort ----------------
def median_of_three(arr, low, high):
    mid = (low + high) // 2

    a = arr[low]
    b = arr[mid]
    c = arr[high]

    # Find median value and return its index
    if (a <= b <= c) or (c <= b <= a):
        return mid
    elif (b <= a <= c) or (c <= a <= b):
        return low
    else:
        return high


def median_quick_sort(arr, low, high, counter):
    if low < high:
        p = median_partition(arr, low, high, counter)
        median_quick_sort(arr, low, p - 1, counter)
        median_quick_sort(arr, p + 1, high, counter)


def median_partition(arr, low, high, counter):
    pivot_index = median_of_three(arr, low, high)

    # Move median pivot to first position
    arr[low], arr[pivot_index] = arr[pivot_index], arr[low]

    return partition_first(arr, low, high, counter)


# ---------------- Testing ----------------
def run_test(name, data):
    print("\nArray Type:", name)

    algorithms = [
        ("Deterministic Quick Sort", deterministic_quick_sort),
        ("Randomized Quick Sort", randomized_quick_sort),
        ("Median-of-Three Quick Sort", median_quick_sort)
    ]

    for algo_name, algo in algorithms:
        arr = data.copy()
        counter = Counter()

        algo(arr, 0, len(arr) - 1, counter)

        print(f"{algo_name}: {counter.comparisons} comparisons")


# ---------------- Main Program ----------------
if __name__ == "__main__":

    n = 10000

    sorted_array = list(range(n))

    reverse_array = list(range(n, 0, -1))

    random_array = [random.randint(0, 100000) for _ in range(n)]

    run_test("Sorted Array", sorted_array)

    run_test("Reverse Sorted Array", reverse_array)

    run_test("Random Array", random_array)
```
