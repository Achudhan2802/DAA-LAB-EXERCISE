```python
import random
import math
import sys

sys.setrecursionlimit(20000)


# ---------------- Comparison Counter ----------------
class Counter:
    def __init__(self):
        self.comparisons = 0


# ---------------- Standard Quick Sort ----------------
def quick_sort(arr, low, high, counter):
    if low < high:
        p = partition(arr, low, high, counter)
        quick_sort(arr, low, p - 1, counter)
        quick_sort(arr, p + 1, high, counter)


def partition(arr, low, high, counter):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        counter.comparisons += 1
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


# ---------------- Heap Sort ----------------
def heap_sort(arr, low, high, counter):
    n = high - low + 1

    # Build heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i, low, counter)

    # Extract elements
    for i in range(n - 1, 0, -1):
        arr[low], arr[low + i] = arr[low + i], arr[low]
        heapify(arr, i, 0, low, counter)


def heapify(arr, n, i, offset, counter):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n:
        counter.comparisons += 1
        if arr[offset + left] > arr[offset + largest]:
            largest = left

    if right < n:
        counter.comparisons += 1
        if arr[offset + right] > arr[offset + largest]:
            largest = right

    if largest != i:
        arr[offset + i], arr[offset + largest] = (
            arr[offset + largest],
            arr[offset + i]
        )
        heapify(arr, n, largest, offset, counter)


# ---------------- Insertion Sort for Small Arrays ----------------
def insertion_sort(arr, low, high, counter):
    for i in range(low + 1, high + 1):
        key = arr[i]
        j = i - 1

        while j >= low:
            counter.comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                j -= 1
            else:
                break

        arr[j + 1] = key


# ---------------- IntroSort ----------------
def introsort(arr, low, high, counter, depth_limit):
    size = high - low + 1

    # Use insertion sort for small arrays
    if size <= 16:
        insertion_sort(arr, low, high, counter)
        return

    # Switch to heap sort if recursion depth exceeded
    if depth_limit == 0:
        heap_sort(arr, low, high, counter)
        return

    # Continue with quick sort
    p = partition(arr, low, high, counter)

    introsort(arr, low, p - 1, counter, depth_limit - 1)
    introsort(arr, p + 1, high, counter, depth_limit - 1)


def intro_sort(arr, counter):
    n = len(arr)
    depth_limit = 2 * int(math.log2(n))

    introsort(arr, 0, n - 1, counter, depth_limit)


# ---------------- Testing ----------------
def test_sort(name, data):
    print("\nArray Type:", name)

    # Standard Quick Sort
    arr1 = data.copy()
    counter1 = Counter()

    quick_sort(arr1, 0, len(arr1) - 1, counter1)

    print("Quick Sort Comparisons :", counter1.comparisons)

    # IntroSort
    arr2 = data.copy()
    counter2 = Counter()

    intro_sort(arr2, counter2)

    print("IntroSort Comparisons  :", counter2.comparisons)


# ---------------- Main ----------------
if __name__ == "__main__":

    n = 10000

    sorted_array = list(range(n))

    random_array = [
        random.randint(0, 100000)
        for _ in range(n)
    ]

    test_sort("Sorted Array", sorted_array)

    test_sort("Random Array", random_array)
```
