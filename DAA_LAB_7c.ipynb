```python
N = 8

# Possible moves of a knight
move_x = [2, 1, -1, -2, -2, -1, 1, 2]
move_y = [1, 2, 2, 1, -1, -2, -2, -1]


def is_safe(x, y, board):
    return 0 <= x < N and 0 <= y < N and board[x][y] == -1


def solve_knight_tour():
    board = [[-1 for _ in range(N)] for _ in range(N)]

    # Start from (0,0)
    board[0][0] = 0

    if solve_util(0, 0, 1, board):
        print("Knight's Tour Solution:\n")
        print_board(board)
    else:
        print("No solution exists.")


def solve_util(x, y, move_no, board):
    if move_no == N * N:
        return True

    for k in range(8):
        next_x = x + move_x[k]
        next_y = y + move_y[k]

        if is_safe(next_x, next_y, board):
            board[next_x][next_y] = move_no

            if solve_util(next_x, next_y, move_no + 1, board):
                return True

            # Backtrack
            board[next_x][next_y] = -1

    return False


def print_board(board):
    for row in board:
        for cell in row:
            print(f"{cell:2}", end=" ")
        print()


# Main
solve_knight_tour()
```
