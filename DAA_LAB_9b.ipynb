```python
# Cloud Server Resource Allocation using FFD and BFD

def first_fit_decreasing(tasks, capacity):
    # Sort tasks in decreasing order
    tasks = sorted(tasks, reverse=True)

    servers = []
    remaining = []

    for task in tasks:
        placed = False

        for i in range(len(servers)):
            if remaining[i] >= task:
                servers[i].append(task)
                remaining[i] -= task
                placed = True
                break

        if not placed:
            servers.append([task])
            remaining.append(capacity - task)

    return servers


def best_fit_decreasing(tasks, capacity):
    # Sort tasks in decreasing order
    tasks = sorted(tasks, reverse=True)

    servers = []
    remaining = []

    for task in tasks:
        best_index = -1
        min_space = float('inf')

        for i in range(len(servers)):
            if remaining[i] >= task:
                space_left = remaining[i] - task

                if space_left < min_space:
                    min_space = space_left
                    best_index = i

        if best_index != -1:
            servers[best_index].append(task)
            remaining[best_index] -= task
        else:
            servers.append([task])
            remaining.append(capacity - task)

    return servers


def display_result(method, servers):
    print(f"\n{method}")
    print("Number of Servers Required:", len(servers))

    for i, server in enumerate(servers, 1):
        used = sum(server)
        print(
            f"Server {i}: Tasks {server} | "
            f"Used RAM = {used} GB | "
            f"Remaining = {8 - used} GB"
        )


# -----------------------------
# Main Program
# -----------------------------

server_capacity = 8  # GB RAM

tasks = [
    1.5, 2.0, 3.5, 0.5, 4.0,
    1.0, 2.5, 3.0, 1.5, 2.0,
    0.5, 4.0, 1.0, 3.5, 2.0
]

print("Task RAM Requirements:")
print(tasks)

print("\nTotal RAM Required:", sum(tasks), "GB")
print("Minimum possible servers:",
      int((sum(tasks) + server_capacity - 1) // server_capacity))


# Apply FFD and BFD
ffd_servers = first_fit_decreasing(tasks, server_capacity)
bfd_servers = best_fit_decreasing(tasks, server_capacity)


# Display results
display_result("First Fit Decreasing (FFD)", ffd_servers)
display_result("Best Fit Decreasing (BFD)", bfd_servers)
```
