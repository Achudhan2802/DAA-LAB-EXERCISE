```python
import random

# Bin capacity
CAPACITY = 1.0


# Next Fit Decreasing (NFD)
def next_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    current_bin = []
    remaining = CAPACITY

    for item in items:
        if item <= remaining:
            current_bin.append(item)
            remaining -= item
        else:
            bins.append(current_bin)
            current_bin = [item]
            remaining = CAPACITY - item

    if current_bin:
        bins.append(current_bin)

    return bins


# First Fit Decreasing (FFD)
def first_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:
        placed = False

        for i in range(len(bins)):
            if remaining[i] >= item:
                bins[i].append(item)
                remaining[i] -= item
                placed = True
                break

        if not placed:
            bins.append([item])
            remaining.append(CAPACITY - item)

    return bins


# Best Fit Decreasing (BFD)
def best_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:
        best_index = -1
        best_remaining = CAPACITY + 1

        for i in range(len(bins)):
            if remaining[i] >= item:
                space = remaining[i] - item

                if space < best_remaining:
                    best_remaining = space
                    best_index = i

        if best_index != -1:
            bins[best_index].append(item)
            remaining[best_index] -= item
        else:
            bins.append([item])
            remaining.append(CAPACITY - item)

    return bins


# Main simulation
trials = 100
items_count = 50

nfd_sum = 0
ffd_sum = 0
bfd_sum = 0

for i in range(trials):

    # Generate random items from 0.1 to 0.9
    items = []

    for j in range(items_count):
        items.append(round(random.uniform(0.1, 0.9), 2))

    nfd_bins = next_fit_decreasing(items)
    ffd_bins = first_fit_decreasing(items)
    bfd_bins = best_fit_decreasing(items)

    nfd_sum += len(nfd_bins)
    ffd_sum += len(ffd_bins)
    bfd_sum += len(bfd_bins)


# Calculate average bins
print("Bin Packing Comparison")
print("----------------------")
print("Trials:", trials)
print("Items per trial:", items_count)

print("\nAverage number of bins used:")
print("NFD =", round(nfd_sum / trials, 2))
print("FFD =", round(ffd_sum / trials, 2))
print("BFD =", round(bfd_sum / trials, 2))
```
