```python
import random

# Bin capacity
CAPACITY = 1.0


# -----------------------------
# Next Fit Decreasing (NFD)
# -----------------------------
def next_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    current_bin = []
    remaining = CAPACITY

    for item in items:
        if item <= remaining:
            current_bin.append(item)
            remaining -= item
        else:
            bins.append(current_bin)
            current_bin = [item]
            remaining = CAPACITY - item

    if current_bin:
        bins.append(current_bin)

    return bins


# -----------------------------
# First Fit Decreasing (FFD)
# -----------------------------
def first_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:
        placed = False

        for i in range(len(bins)):
            if remaining[i] >= item:
                bins[i].append(item)
                remaining[i] -= item
                placed = True
                break

        if not placed:
            bins.append([item])
            remaining.append(CAPACITY - item)

    return bins


# -----------------------------
# Best Fit Decreasing (BFD)
# -----------------------------
def best_fit_decreasing(items):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:
        best_index = -1
        best_space = float('inf')

        for i in range(len(bins)):
            if remaining[i] >= item:
                space = remaining[i] - item

                if space < best_space:
                    best_space = space
                    best_index = i

        if best_index != -1:
            bins[best_index].append(item)
            remaining[best_index] -= item
        else:
            bins.append([item])
            remaining.append(CAPACITY - item)

    return bins


# -----------------------------
# Run Simulation
# -----------------------------
trials = 100
items_count = 50

nfd_total = 0
ffd_total = 0
bfd_total = 0

for _ in range(trials):

    # Generate 50 random items (0.1 to 0.9)
    items = [
        round(random.uniform(0.1, 0.9), 2)
        for _ in range(items_count)
    ]

    nfd_bins = next_fit_decreasing(items)
    ffd_bins = first_fit_decreasing(items)
    bfd_bins = best_fit_decreasing(items)

    nfd_total += len(nfd_bins)
    ffd_total += len(ffd_bins)
    bfd_total += len(bfd_bins)


# Average bins used
print("Results after", trials, "trials")
print("--------------------------------")
print("Average bins used by NFD :",
      round(nfd_total / trials, 2))

print("Average bins used by FFD :",
      round(ffd_total / trials, 2))

print("Average bins used by BFD :",
      round(bfd_total / trials, 2))
```
