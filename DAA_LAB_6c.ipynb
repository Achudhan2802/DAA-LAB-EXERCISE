```python
import sys

# Example: Chain of 6 matrices
# A1 = 30x35
# A2 = 35x15
# A3 = 15x5
# A4 = 5x10
# A5 = 10x20
# A6 = 20x25
p = [30, 35, 15, 5, 10, 20, 25]
n = len(p) - 1

# -------------------------------
# Top-Down Memoized Approach
# -------------------------------
memo = [[-1] * (n + 1) for _ in range(n + 1)]

def matrix_chain_memo(i, j):
    if i == j:
        return 0

    if memo[i][j] != -1:
        return memo[i][j]

    memo[i][j] = sys.maxsize

    for k in range(i, j):
        cost = (matrix_chain_memo(i, k) +
                matrix_chain_memo(k + 1, j) +
                p[i - 1] * p[k] * p[j])

        memo[i][j] = min(memo[i][j], cost)

    return memo[i][j]

# -------------------------------
# Bottom-Up DP Approach
# -------------------------------
dp = [[0] * (n + 1) for _ in range(n + 1)]

for length in range(2, n + 1):
    for i in range(1, n - length + 2):
        j = i + length - 1
        dp[i][j] = sys.maxsize

        for k in range(i, j):
            cost = (dp[i][k] +
                    dp[k + 1][j] +
                    p[i - 1] * p[k] * p[j])

            dp[i][j] = min(dp[i][j], cost)

# -------------------------------
# Results
# -------------------------------
top_down_cost = matrix_chain_memo(1, n)
bottom_up_cost = dp[1][n]

print("Top-Down (Memoized) Cost :", top_down_cost)
print("Bottom-Up DP Cost        :", bottom_up_cost)

if top_down_cost == bottom_up_cost:
    print("Verification: Both methods give the SAME minimum cost.")
else:
    print("Verification: Costs are DIFFERENT.")
```
