```python
from itertools import permutations

INF = float('inf')

# -------- 6-City Distance Matrix (Road Distances in km) --------
cost = [
    [INF, 10, 15, 20, 25, 18],
    [10, INF, 35, 25, 17, 28],
    [15, 35, INF, 30, 28, 16],
    [20, 25, 30, INF, 22, 14],
    [25, 17, 28, 22, INF, 24],
    [18, 28, 16, 14, 24, INF]
]

cities = ["A", "B", "C", "D", "E", "F"]
n = len(cost)


# -------- Brute Force TSP --------
def tsp_brute_force(cost):
    cities_list = list(range(1, n))
    best_cost = INF
    best_path = None

    for perm in permutations(cities_list):
        path = [0] + list(perm) + [0]

        total = 0
        for i in range(n):
            total += cost[path[i]][path[i + 1]]

        if total < best_cost:
            best_cost = total
            best_path = path

    return best_path, best_cost


# -------- Nearest Neighbour Heuristic --------
def nearest_neighbour(cost):
    visited = [False] * n
    path = [0]
    visited[0] = True
    total_cost = 0
    current = 0

    for _ in range(n - 1):
        nearest = -1
        min_cost = INF

        for city in range(n):
            if not visited[city] and cost[current][city] < min_cost:
                min_cost = cost[current][city]
                nearest = city

        path.append(nearest)
        visited[nearest] = True
        total_cost += min_cost
        current = nearest

    path.append(0)
    total_cost += cost[current][0]

    return path, total_cost


# -------- Display Distance Matrix --------
print("6-City Distance Matrix (km)\n")

print(f"{'':>4}", end="")
for city in cities:
    print(f"{city:>5}", end="")
print()

for i in range(n):
    print(f"{cities[i]:>4}", end="")
    for value in cost[i]:
        if value == INF:
            print(f"{'INF':>5}", end="")
        else:
            print(f"{value:>5}", end="")
    print()


# -------- Solve TSP --------
bf_path, bf_cost = tsp_brute_force(cost)
nn_path, nn_cost = nearest_neighbour(cost)

print("\nBrute Force (Optimal Solution)")
print("Route :", " -> ".join(cities[i] for i in bf_path))
print("Cost  :", bf_cost, "km")

print("\nNearest Neighbour Heuristic")
print("Route :", " -> ".join(cities[i] for i in nn_path))
print("Cost  :", nn_cost, "km")

print("\nComparison")
if bf_cost == nn_cost:
    print("Nearest Neighbour found the optimal solution.")
else:
    print("Nearest Neighbour did NOT find the optimal solution.")
    print("Difference =", nn_cost - bf_cost, "km")
```
