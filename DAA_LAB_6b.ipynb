```python
import sys

# Function to print the optimal parenthesization
def print_optimal(s, i, j):
    if i == j:
        print(f"A{i}", end="")
    else:
        print("(", end="")
        print_optimal(s, i, s[i][j])
        print_optimal(s, s[i][j] + 1, j)
        print(")", end="")

# Matrix dimensions:
# T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)
p = [4, 4, 4, 4, 10000, 1]
n = len(p) - 1

# DP tables
m = [[0] * (n + 1) for _ in range(n + 1)]
s = [[0] * (n + 1) for _ in range(n + 1)]

# Matrix Chain Multiplication
for L in range(2, n + 1):          # Chain length
    for i in range(1, n - L + 2):
        j = i + L - 1
        m[i][j] = sys.maxsize

        for k in range(i, j):
            cost = (m[i][k] +
                    m[k + 1][j] +
                    p[i - 1] * p[k] * p[j])

            if cost < m[i][j]:
                m[i][j] = cost
                s[i][j] = k

print("Minimum Scalar Multiplications =", m[1][n])
print("Optimal Parenthesization =", end=" ")
print_optimal(s, 1, n)
print()
```
