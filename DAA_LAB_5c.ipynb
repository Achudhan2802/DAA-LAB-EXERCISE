```python
import random


def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


def quick_select(arr, low, high, k):
    if low <= high:

        pivot_index = partition(arr, low, high)

        # Pivot is the k-th smallest element
        if pivot_index == k:
            return arr[pivot_index]

        # Search left half
        elif pivot_index > k:
            return quick_select(arr, low, pivot_index - 1, k)

        # Search right half
        else:
            return quick_select(arr, pivot_index + 1, high, k)


# ---------------- Main Program ----------------

# Generate 1000 random integers
n = 1000
arr = [random.randint(1, 10000) for _ in range(n)]

print("Array Size :", n)

# k = 1 (Minimum)
minimum = quick_select(arr.copy(), 0, n - 1, 0)

# k = n (Maximum)
maximum = quick_select(arr.copy(), 0, n - 1, n - 1)

# k = n//2 (Median)
median = quick_select(arr.copy(), 0, n - 1, n // 2)

print("Minimum Element :", minimum)
print("Maximum Element :", maximum)
print("Median Element  :", median)
```
