```python
import random

comparison_count = 0


def find_four(arr, low, high):
    global comparison_count

    # Base case: one element
    if low == high:
        return arr[low], None, arr[low], None

    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2

    lmin, lsmin, lmax, lsmax = find_four(arr, low, mid)
    rmin, rsmin, rmax, rsmax = find_four(arr, mid + 1, high)

    # -------- Find Minimum --------
    comparison_count += 1
    if lmin < rmin:
        overall_min = lmin

        if rsmin is None:
            second_min = rmin
        else:
            comparison_count += 1
            second_min = lsmin if lsmin is not None and lsmin < rmin else rmin
    else:
        overall_min = rmin

        if lsmin is None:
            second_min = lmin
        else:
            comparison_count += 1
            second_min = rsmin if rsmin is not None and rsmin < lmin else lmin

    # -------- Find Maximum --------
    comparison_count += 1
    if lmax > rmax:
        overall_max = lmax

        if rsmax is None:
            second_max = rmax
        else:
            comparison_count += 1
            second_max = lsmax if lsmax is not None and lsmax > rmax else rmax
    else:
        overall_max = rmax

        if lsmax is None:
            second_max = lmax
        else:
            comparison_count += 1
            second_max = rsmax if rsmax is not None and rsmax > lmax else lmax

    return overall_min, second_min, overall_max, second_max


# ---------------- Main Program ----------------

arr = [12, 5, 8, 20, 15, 2, 25, 18, 10, 30]

comparison_count = 0

minimum, second_minimum, maximum, second_maximum = find_four(
    arr, 0, len(arr) - 1)

print("Array:", arr)
print("Minimum :", minimum)
print("Second Minimum :", second_minimum)
print("Maximum :", maximum)
print("Second Maximum :", second_maximum)
print("Total Comparisons :", comparison_count)

# ---------------- Performance ----------------

print("\n{:>8} {:>18}".format("Size", "Comparisons"))
print("-" * 30)

for size in [10, 100, 1000, 10000]:
    arr = random.sample(range(1, size * 10), size)

    comparison_count = 0
    find_four(arr, 0, len(arr) - 1)

    print("{:>8} {:>18}".format(size, comparison_count))
```
